```python
import pandas as pd
```

```python
data={ "Name":["Rahul","Mbappe","Cristiano"],
      "Marks":[11,22,7],
      "Branch":["CSE","ECE","FOOTBALL"]}
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
print(df["Marks"])
```

```python
print(df["Name"])
```

```python
print("First row using row:")
print(df.iloc[0])
```

```python
print("First person marks:",df.iloc[0,1])
```

```python
print("Row slicing of first two rows:")
print(df.iloc[0:2])
```

```python
print("Both row and coln slicing")
print(df.iloc[0:3,[0,1]])
```

```python
print("Filtering:")
res=df[df["Marks"]>9]
print("Students with marks greater than 9:")
print(res)
```
